```python
import os
import zipfile

# Caminho da pasta que você quer compactar
pasta_origem = "/content/drive/MyDrive/Dataset Base/SpotBID Set"

# Caminho onde o arquivo .zip será salvo
zip_destino = "/content/drive/MyDrive/Dataset Base/SpotBID_Set.zip"

# Função para compactar recursivamente
def zip_folder(folder_path, output_path):
    with zipfile.ZipFile(output_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for root, dirs, files in os.walk(folder_path):
            for file in files:
                file_path = os.path.join(root, file)
                arcname = os.path.relpath(file_path, start=folder_path)
                zipf.write(file_path, arcname)
    print(f"Pasta '{folder_path}' compactada em '{output_path}'")

# Executar
zip_folder(pasta_origem, zip_destino)
```

```python
import zipfile
import os
import shutil
from tqdm import tqdm

# Caminho do arquivo ZIP
zip_path = '/content/drive/MyDrive/Dataset Base/SpotBID_Set.zip'
extract_dir = '/content/SpotBID_Extracted'
destino = '/content/SpotBID_Local'

# Extrair o ZIP
print("Extraindo o ZIP...")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)
print("Extração concluída.")

# Remove pasta de destino antiga, se existir
if os.path.exists(destino):
    shutil.rmtree(destino)
os.makedirs(destino)

# Cópia recursiva com barra de progresso
print("Copiando arquivos extraídos...")
for root, dirs, files in os.walk(extract_dir):
    for file in tqdm(files, desc="Transferindo"):
        origem_path = os.path.join(root, file)

        # Recria a subpasta no destino
        rel_path = os.path.relpath(root, extract_dir)
        destino_dir = os.path.join(destino, rel_path)
        os.makedirs(destino_dir, exist_ok=True)

        # Copia o arquivo
        destino_path = os.path.join(destino_dir, file)
        shutil.copy2(origem_path, destino_path)

print("Arquivos organizados em /content/SpotBID_Local.")
```

```python
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import BinaryCrossentropy
from tensorflow.keras.utils import Sequence
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping

# Pastas
train_dir = "/content/SpotBID_Local/train_folder"
val_dir = "/content/SpotBID_Local/validation_folder"
checkpoint_dir = "/content/drive/MyDrive/detector_documento_spotbid_checkpoints"
os.makedirs(checkpoint_dir, exist_ok=True)

# Configurações
IMG_SIZE = 256
BATCH_SIZE = 8
EPOCHS = 30
INITIAL_EPOCH = 5


# DataGenerator
class SpotBIDGenerator(Sequence):
    def __init__(self, folder, batch_size=8, img_size=256, shuffle=True):
        self.folder = folder
        self.batch_size = batch_size
        self.img_size = img_size
        self.shuffle = shuffle

        self.image_paths = sorted([os.path.join(folder, f) for f in os.listdir(folder) if f.endswith('_in.png')])
        self.mask_paths = sorted([os.path.join(folder, f) for f in os.listdir(folder) if f.endswith('_gt.png')])

        self.indices = np.arange(len(self.image_paths))
        if self.shuffle:
            np.random.shuffle(self.indices)

    def __len__(self):
        return int(np.ceil(len(self.image_paths) / self.batch_size))

    def __getitem__(self, idx):
        batch_indices = self.indices[idx * self.batch_size:(idx + 1) * self.batch_size]
        batch_x, batch_y = [], []

        for i in batch_indices:
            img = cv2.imread(self.image_paths[i])
            img = cv2.resize(img, (self.img_size, self.img_size)) / 255.0

            mask = cv2.imread(self.mask_paths[i], cv2.IMREAD_GRAYSCALE)
            mask = cv2.resize(mask, (self.img_size, self.img_size))
            mask = (mask > 127).astype(np.float32)
            mask = np.expand_dims(mask, axis=-1)

            batch_x.append(img)
            batch_y.append(mask)

        return np.array(batch_x), np.array(batch_y)

    def on_epoch_end(self):
        if self.shuffle:
            np.random.shuffle(self.indices)

# Geradores
train_gen = SpotBIDGenerator(train_dir, batch_size=BATCH_SIZE, img_size=IMG_SIZE)
val_gen = SpotBIDGenerator(val_dir, batch_size=BATCH_SIZE, img_size=IMG_SIZE, shuffle=False)

# Modelo U-Net
def build_model(input_shape=(256, 256, 3)):
    inputs = Input(input_shape)
    c1 = Conv2D(64, (3, 3), activation='relu', padding='same')(inputs)
    c1 = Conv2D(64, (3, 3), activation='relu', padding='same')(c1)
    p1 = MaxPooling2D((2, 2))(c1)

    c2 = Conv2D(128, (3, 3), activation='relu', padding='same')(p1)
    c2 = Conv2D(128, (3, 3), activation='relu', padding='same')(c2)
    p2 = MaxPooling2D((2, 2))(c2)

    c3 = Conv2D(256, (3, 3), activation='relu', padding='same')(p2)
    c3 = Conv2D(256, (3, 3), activation='relu', padding='same')(c3)

    u1 = UpSampling2D((2, 2))(c3)
    concat1 = concatenate([u1, c2])
    c4 = Conv2D(128, (3, 3), activation='relu', padding='same')(concat1)
    c4 = Conv2D(128, (3, 3), activation='relu', padding='same')(c4)

    u2 = UpSampling2D((2, 2))(c4)
    concat2 = concatenate([u2, c1])
    c5 = Conv2D(64, (3, 3), activation='relu', padding='same')(concat2)
    c5 = Conv2D(64, (3, 3), activation='relu', padding='same')(c5)

    outputs = Conv2D(1, (1, 1), activation='sigmoid')(c5)
    return Model(inputs, outputs)

# Callbacks
checkpoint_cb = ModelCheckpoint(
    filepath=os.path.join(checkpoint_dir, "model_epoch_{epoch:02d}.h5"),
    save_weights_only=False,
    save_freq='epoch',
    verbose=1
)

earlystop_cb = EarlyStopping(
    monitor='val_loss',
    patience=6,
    restore_best_weights=True
)

# Modelo
model = build_model()
model.compile(optimizer=Adam(), loss=BinaryCrossentropy(), metrics=['accuracy'])
model.load_weights('/content/drive/MyDrive/detector_documento_spotbid_checkpoints/model_epoch_05.h5')

# Treinamento
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=EPOCHS,
    initial_epoch=INITIAL_EPOCH,
    callbacks=[checkpoint_cb, earlystop_cb]
)

# Gráficos
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Perda - treino')
plt.plot(history.history['val_loss'], label='Perda - validação')
plt.title('Loss por época')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Acurácia - treino')
plt.plot(history.history['val_accuracy'], label='Acurácia - validação')
plt.title('Acurácia por época')
plt.xlabel('Épocas')
plt.ylabel('Acurácia')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
from sklearn.metrics import accuracy_score, jaccard_score
from tqdm import tqdm

# Caminhos
test_folder = "/content/drive/MyDrive/Datasets Teste/SpotBID Teste Final"
model_path = "/content/drive/MyDrive/detector_documento_spotbid_checkpoints/model_epoch_21.h5"

# Parâmetros
IMG_SIZE = 256

# Carregar modelo
model = load_model(model_path)
print("Modelo carregado com sucesso!")

# Função para carregar e preprocessar
def load_test_data(folder):
    images = sorted([f for f in os.listdir(folder) if f.endswith("_in.png")])
    masks = sorted([f for f in os.listdir(folder) if f.endswith("_gt.png")])

    X, Y = [], []
    for img_file, mask_file in tqdm(zip(images, masks), total=len(images), desc="🔄 Carregando imagens"):
        img_path = os.path.join(folder, img_file)
        mask_path = os.path.join(folder, mask_file)

        img = cv2.imread(img_path)
        img = cv2.resize(img, (IMG_SIZE, IMG_SIZE)) / 255.0

        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        mask = cv2.resize(mask, (IMG_SIZE, IMG_SIZE))
        mask = (mask > 127).astype(np.uint8)

        X.append(img)
        Y.append(mask)

    return np.array(X), np.array(Y)

# Carrega imagens e máscaras
X_test, Y_true = load_test_data(test_folder)

# Previsão
pred_masks = model.predict(X_test, verbose=1)
pred_masks = (pred_masks.squeeze() > 0.5).astype(np.uint8)
Y_true = Y_true.squeeze()

# Cálculo das métricas
iou = jaccard_score(Y_true.flatten(), pred_masks.flatten())
acc = accuracy_score(Y_true.flatten(), pred_masks.flatten())

print(f"\nAvaliação no SpotBID Teste Final:")
print(f"Acurácia Pixel a Pixel: {acc:.4f}")
print(f"IoU (Jaccard):          {iou:.4f}")
```

```python
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import BinaryCrossentropy
from tensorflow.keras.utils import Sequence
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping

# Pastas
train_dir = "/content/SpotBID_Local/train_folder"
val_dir = "/content/SpotBID_Local/validation_folder"
save_path = "/content/drive/MyDrive/modelo_unet_novo.h5"

# Configurações
IMG_SIZE = 256
BATCH_SIZE = 8
EPOCHS = 30

#  DataGenerator
class SpotBIDGenerator(Sequence):
    def __init__(self, folder, batch_size=8, img_size=256, shuffle=True):
        self.folder = folder
        self.batch_size = batch_size
        self.img_size = img_size
        self.shuffle = shuffle

        self.image_paths = sorted([os.path.join(folder, f) for f in os.listdir(folder) if f.endswith('_in.png')])
        self.mask_paths = sorted([os.path.join(folder, f) for f in os.listdir(folder) if f.endswith('_gt.png')])

        self.indices = np.arange(len(self.image_paths))
        if self.shuffle:
            np.random.shuffle(self.indices)

    def __len__(self):
        return int(np.ceil(len(self.image_paths) / self.batch_size))

    def __getitem__(self, idx):
        batch_indices = self.indices[idx * self.batch_size:(idx + 1) * self.batch_size]
        batch_x, batch_y = [], []

        for i in batch_indices:
            img = cv2.imread(self.image_paths[i])
            img = cv2.resize(img, (self.img_size, self.img_size)) / 255.0

            mask = cv2.imread(self.mask_paths[i], cv2.IMREAD_GRAYSCALE)
            mask = cv2.resize(mask, (self.img_size, self.img_size))
            mask = (mask > 127).astype(np.float32)
            mask = np.expand_dims(mask, axis=-1)

            batch_x.append(img)
            batch_y.append(mask)

        return np.array(batch_x), np.array(batch_y)

    def on_epoch_end(self):
        if self.shuffle:
            np.random.shuffle(self.indices)

#  Geradores
train_gen = SpotBIDGenerator(train_dir, batch_size=BATCH_SIZE, img_size=IMG_SIZE)
val_gen = SpotBIDGenerator(val_dir, batch_size=BATCH_SIZE, img_size=IMG_SIZE, shuffle=False)

# Modelo U-Net
def build_model(input_shape=(256, 256, 3)):
    inputs = Input(input_shape)
    c1 = Conv2D(64, (3, 3), activation='relu', padding='same')(inputs)
    c1 = Conv2D(64, (3, 3), activation='relu', padding='same')(c1)
    p1 = MaxPooling2D((2, 2))(c1)

    c2 = Conv2D(128, (3, 3), activation='relu', padding='same')(p1)
    c2 = Conv2D(128, (3, 3), activation='relu', padding='same')(c2)
    p2 = MaxPooling2D((2, 2))(c2)

    c3 = Conv2D(256, (3, 3), activation='relu', padding='same')(p2)
    c3 = Conv2D(256, (3, 3), activation='relu', padding='same')(c3)

    u1 = UpSampling2D((2, 2))(c3)
    concat1 = concatenate([u1, c2])
    c4 = Conv2D(128, (3, 3), activation='relu', padding='same')(concat1)
    c4 = Conv2D(128, (3, 3), activation='relu', padding='same')(c4)

    u2 = UpSampling2D((2, 2))(c4)
    concat2 = concatenate([u2, c1])
    c5 = Conv2D(64, (3, 3), activation='relu', padding='same')(concat2)
    c5 = Conv2D(64, (3, 3), activation='relu', padding='same')(c5)

    outputs = Conv2D(1, (1, 1), activation='sigmoid')(c5)
    return Model(inputs, outputs)

# Compilação
model = build_model()
model.compile(optimizer=Adam(learning_rate=1e-3), loss=BinaryCrossentropy(), metrics=['accuracy'])

#  Callbacks
checkpoint_cb = ModelCheckpoint(
    filepath=save_path,
    save_weights_only=False,
    save_best_only=True,
    monitor='val_loss',
    verbose=1
)

earlystop_cb = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

# Treinamento
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[checkpoint_cb, earlystop_cb]
)
```
